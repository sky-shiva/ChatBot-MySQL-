```python
import fitz
import easyocr
import os



PDF_PATH = "D:\HACKATHONS and PROJECTS\Project ChatBot(SQL)\sql notes.pdf"

IMAGE_FOLDER = "page_images"
TEXT_FOLDER = "extracted_text"

os.makedirs(IMAGE_FOLDER, exist_ok=True)
os.makedirs(TEXT_FOLDER, exist_ok=True)

# -----------------------------
# LOAD OCR
# -----------------------------

reader = easyocr.Reader(['en'])

full_text = ""

# -----------------------------
# OPEN PDF
# -----------------------------

doc = fitz.open(PDF_PATH)

for page_num in range(len(doc)):

    page = doc.load_page(page_num)

    pix = page.get_pixmap()

    image_path = f"{IMAGE_FOLDER}/page_{page_num}.png"

    pix.save(image_path)

    # OCR
    result = reader.readtext(image_path, detail=0)

    page_text = "\n".join(result)

    full_text += f"\n--- PAGE {page_num} ---\n"
    full_text += page_text

    print(f"Processed page {page_num}")

# -----------------------------
# SAVE TEXT
# -----------------------------

output_path = f"{TEXT_FOLDER}/output.txt"

with open(output_path, "w", encoding="utf-8") as f:
    f.write(full_text)

print("\nOCR COMPLETED!")
print(f"Saved at: {output_path}")
```
